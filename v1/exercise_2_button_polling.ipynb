# Exercise 2: Button Polling & Multi-Note Inputs

In embedded systems, **button polling** is the technique of repeatedly checking the voltage state of digital input pins inside our main execution loop. Just like reading paddle movement buttons in a video game, our synthesizer polls physical pushbuttons to know when to trigger a musical note.

## Starter Code Walkthrough (`SYNTH_BASE.ino`)

1. Unzip `SYNTH_BASE.zip` and open `SYNTH_BASE.ino` in the Arduino IDE (do not edit the `.cpp` or `.h` helper files during this exercise).
2. Upload the starter sketch to your **Seeeduino Nano**.
3. Press the buttons on your PCB to find which single button currently triggers a note.

Inspect how that single note is wired in `SYNTH_BASE.ino` across five sections:

### Definitions


````python
// Line 22: Constant mapping the note trigger to Digital Pin 7
#define ADSR_NOTE_0 7

// Line 39: Boolean storing the button state (HIGH = unpressed)
bool notePinState0 = HIGH;
````

### Setup Code


````python
// Line 60: Enable the internal pull-up resistor so unpressed reads HIGH
pinMode(ADSR_NOTE_0, INPUT_PULLUP);
````

### Loop & Input Polling Code


````python
// Inside checkInputs() (Lines 91-92): Poll the pin state
case 0:
    notePinState0 = digitalRead(ADSR_NOTE_0);
    break;

// Inside loop() (Line 122): Pass polled state(s) to the audio engine
updateADSR(notePinState0);
````
