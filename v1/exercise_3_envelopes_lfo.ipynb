# Exercise 3: Envelopes (ADSR) & LFO Modulation

A raw oscillator running at constant amplitude sounds like an alarm clock. Real instruments shape how a sound evolves over time—a piano hammer strikes sharply and fades, while a violin swells gradually. Synthesizers replicate this using **Envelopes** and **Low-Frequency Oscillators (LFOs)**.

## The Two Synthesizer Modes

With your 6-note code from Exercise 2 running on the PCB, press the **Mode Switch Button** to toggle between the two control states (monitor the Serial Monitor if needed to see which state is active):

### 1. ADSR State (Amplitude Envelope)
Controls how a note's volume behaves from the moment you press a button until after you release it:
* **Attack:** How fast the sound ramps up from silence to peak volume after pressing a key.
* **Decay:** How fast the volume drops from peak level down to the sustain level.
* **Sustain:** The steady volume level held while you keep the button pressed.
* **Release:** How long it takes for the sound to fade out after you let go of the button.

### 2. LFO State (Filter & Oscillator Modulation)
Uses a slow, sub-audio wave (an **LFO**) to automatically sweep synthesizer parameters:
* **Waveform:** Selects the base wave shape (Sine, Triangle, Sawtooth, Square).
* **Frequency / Pitch:** Shifts the base tuning of your keyboard.
* **Cutoff / Intensity:** Controls the low-pass filter threshold (muffling vs. brightening the tone).
* **LFO Cutoff Rate:** Controls how fast the filter automatically wobbles.
* **Resonance:** Boosts frequencies right at the cutoff point, creating a vocal or "squelchy" sci-fi quality.
* **Amplitude:** Master output volume.

## Challenge: Sound Design Lab

Use your 6-key keyboard and mode toggle button to map out which physical potentiometer controls each parameter above, then dial in two contrasting patches:

1. **Patch A — Percussive Pluck / Drum:** Configure the ADSR knobs so pressing a key produces a sharp, short hit with zero lingering hold even if you keep your finger on the button. *(Hint: Think about Attack, Decay, and Sustain levels.)*
2. **Patch B — Swelling Sci-Fi Pad:** Configure the ADSR and LFO knobs so notes fade in slowly, echo after you release the button, and wobble with high filter resonance.
