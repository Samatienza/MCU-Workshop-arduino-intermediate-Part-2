# Exercise 1: Digital Waveforms & Acoustic Physics

To program a synthesizer, we first need to understand how computers represent sound. Sound travels as pressure waves characterized by two core properties:

* **Frequency ($f$):** The number of wave oscillations per second, measured in **Hertz (Hz)**. Our ears perceive frequency as **pitch**.
* **Amplitude ($A$):** The maximum displacement of the wave from equilibrium. Our ears perceive amplitude as **volume** (measured logarithmically in decibels, $\text{dB}$, where $+10\text{ dB}$ doubles perceived loudness).

Mathematically, a pure tone is modeled as a sine wave:
$$y = A\sin\left[B(x - C)\right] + D$$
where the period of one full cycle is $T = \frac{2\pi}{B} = \frac{1}{f}$.

## Part 1: Pitch, Amplitude & Octaves in Audacity

1. Open **Audacity** with an empty project. Navigate to **Generate** $\rightarrow$ **Tone**.
2. Generate a **Sine** wave with Frequency = **`440 Hz`**, Amplitude = **`0.8`**, and Duration = **`2 seconds`**. Zoom in until you clearly see the individual oscillations.
3. Unselect the track, right-click the empty gray area below it, and select **Add Mono Track**.
4. On the new track, generate a **Sine** wave with Frequency = **`523 Hz`**, Amplitude = **`0.6`**, and Duration = **`2 seconds`**.
5. Drag the `523 Hz` clip onto the top track so it plays from **2s to 4s**, delete the empty bottom track, and press **Play**.

## Part 2: Superimposition — Harmony vs. Acoustic Beats

When two sound waves play simultaneously, their amplitudes add together (**superimposition**).

1. Clear your tracks. Create two separate mono tracks and generate:
   * Track 1: **Sine**, **`523 Hz`** ($C_5$), Amplitude **`0.5`**, Duration **`2s`**
   * Track 2: **Sine**, **`784 Hz`** ($G_5$), Amplitude **`0.5`**, Duration **`2s`**
2. Play them together. Because their frequencies form a clean musical interval (a perfect fifth), they blend into a pleasant **harmony**.
3. Now delete both tracks and generate two closely spaced frequencies on separate tracks:
   * Track 1: **Sine**, **`300 Hz`**, Amplitude **`1.0`**, Duration **`2s`**
   * Track 2: **Sine**, **`305 Hz`**, Amplitude **`1.0`**, Duration **`2s`**
4. Play them together. Instead of harmony, you will hear a rhythmic pulsing wobble called **acoustic beats** ($f_{\text{beat}} = |f_1 - f_2| = 5\text{ Hz}$), caused by alternating constructive and destructive interference.

## Part 3: Waveform Shapes & Timbre

Synthesizers create rich instrument textures (**timbre**) by using non-sinusoidal waveforms:

1. Clear your project and place three **`440 Hz`** tones back-to-back on a single track (2 seconds each):
   * **0s – 2s:** **Sine** wave (Amplitude `0.8`)
   * **2s – 4s:** **Triangle** wave (Amplitude `0.85`)
   * **4s – 6s:** **Square** wave (Amplitude `0.8`)
2. Listen to all three. Notice that while the pitch ($440\text{ Hz}$) never changes, the **Square wave** sounds much harsher and buzzier (popular in EDM basslines) due to its sharp vertical jumps and rich odd harmonics.
